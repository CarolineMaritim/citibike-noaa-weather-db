```python
# import modules
import pandas as pd
import json
```

```python
# load data
df = pd.read_csv('newark_airport_2016.csv')

display(df.head(1))
```

```python
# drop other columns remain with core values DATE, PRCP, SNOW,SNWD,TMAX,TMIN

print(df.columns)
df = df.drop(['AWND', 'STATION', 'NAME', 'PGTM', 'TSUN', 'WDF2', 'WDF5', 'WSF2', 'WSF5'], axis=1, errors = 'ignore')

display(df.head(2))
```

```python
# Rename columns

df.rename(columns={'DATE':'date','PRCP':'precipitation', 'SNOW':'snow_fall', 'SNWD':'snow_depth', 'TAVG':'avg_temp', 'TMAX':'max_temp', 'TMIN':'min_temp'}, inplace=True)

display(df.head(2))
```

```python
# save cleaned data
df.to_csv('weather_clean.csv', index=False)
print(df.isnull().sum())
```
