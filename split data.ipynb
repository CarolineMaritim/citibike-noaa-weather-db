```python
# separate data to bike, person, station and trip

import pandas as pd

df = pd.read_csv('bike_trip_clean.csv')

bike = df['bike_id']
bike.to_csv('bike.csv', index=False)

person = df[['user_type', 'gender', 'birth_year', 'bike_id']]
person.to_csv('person.csv', index=False)

station_start = df[['start_station_id','start_station_name','bike_id','start_station_lat','start_station_long']]
station_start.to_csv('station_start.csv', index=False)

station_end = df[['end_station_id','end_station_name','bike_id','end_station_lat','end_station_long']]
station_end.to_csv('station_end.csv', index=False)


trip = df[['bike_id','start_date', 'start_time1', 'end_date', 'end_time1', 'time_sec', 'start_station_name', 'start_station_id','end_station_name', 'end_station_id']]
trip.to_csv('trip.csv', index=False)

# bike_dup = bike.bike_id.duplicated().count()
```
